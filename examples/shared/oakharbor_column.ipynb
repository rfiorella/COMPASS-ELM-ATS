```python
import numpy as np
import watershed_workflow.mesh
```

```python
z_vert = np.array([5.5, 5.48248718208374, 5.45490821282407, 5.40943817958166, 
    5.33447076859545, 5.21087040349317, 5.00708785248273, 4.6711072260343, 
    4.11716882066562, 3.20387878907656, 1.69811808767728, -0.784461609304053, 
    -4.87754356192545, -11.6258948399312, -22.7520451341356, 
    -36.6031972760992,])

dzs = z_vert[:-1] - z_vert[1:]
print(len(dzs))
print(dzs)
```

## Oak Harbor Column

```python
x = np.array([0,100])
z = np.array([5.5, 5.5])
m2 = watershed_workflow.mesh.Mesh2D.from_Transect(x, z, width=100)
```

## Oak Harbor transect

The full transect uses lat-long to match columns and ensure columns align.  In a real run, we would project our mesh coordinates into lat-long and save them.  For this synthetic transect, we prescribe lat-lon.

Note that to use this mode, we also have to pre-partition.  

```python
x = np.linspace(0, 500, 6)
z = np.linspace(5, 0, 6)
print(x,z)
m2_transect2 = watershed_workflow.mesh.Mesh2D.from_Transect(x, z, width=100)
```

```python
# set the lat and lon and add it to data so that it gets partitioned
lat = [41.7505, 41.7505, 41.7505, 41.7505, 41.7505]
lon = [276.7505, 276.7515, 276.7525, 276.7535, 276.7545]
m2_transect2.cell_data['latitude'] = lat
m2_transect2.cell_data['longitude'] = lon
```

```python
# partition the m2 mesh
m2_transect2 = m2_transect2.partition(2)

# write the m2 data to file
m2_transect2.writeCellData('oakharbor_transect.h5', ['longitude', 'latitude'])
```

```python
m3_transect2 = watershed_workflow.mesh.Mesh3D.extruded_Mesh2D(m2_transect2, 'constant', dzs, [1,]*15, 
                                                    list(range(1001, 1011))+[1011,]*5)
```

```python
m3_transect2.writeExodus('oakharbor_transect.exo', 'one block')
```
